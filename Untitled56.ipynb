```python
# Create a copy of the original dataset
cleaned_df = df.copy()

# Fill missing Age values with median
cleaned_df["Age"] = cleaned_df["Age"].fillna(cleaned_df["Age"].median())

# Fill missing Embarked values with mode
cleaned_df["Embarked"] = cleaned_df["Embarked"].fillna(
    cleaned_df["Embarked"].mode()[0]
)

# Drop Cabin because of a very high percentage of missing values
cleaned_df = cleaned_df.drop(columns=["Cabin"])

# Check missing values after cleaning
print(cleaned_df.isnull().sum())
```

```python
plt.figure(figsize=(6,4))

sns.countplot(data=cleaned_df, x="Survived")

plt.title("Survival Distribution")
plt.xlabel("Survived (0 = No, 1 = Yes)")
plt.ylabel("Number of Passengers")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(data=cleaned_df, x="Age", bins=30, kde=True)

plt.title("Age Distribution of Passengers")
plt.xlabel("Age")
plt.ylabel("Number of Passengers")

plt.show()
```

```python
plt.figure(figsize=(7,5))

sns.countplot(data=cleaned_df, x="Sex", hue="Survived")

plt.title("Survival by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Passengers")

plt.show()
```

```python
plt.figure(figsize=(7,5))

sns.countplot(data=cleaned_df, x="Pclass", hue="Survived")

plt.title("Survival by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Number of Passengers")

plt.show()
```

```python
plt.figure(figsize=(9,6))

correlation = cleaned_df.select_dtypes(include="number").corr()

sns.heatmap(correlation, annot=True, cmap="coolwarm", fmt=".2f")

plt.title("Correlation Heatmap")
plt.show()
```

```python
gender_survival = cleaned_df.groupby("Sex")["Survived"].mean() * 100

print(gender_survival)
```

```python
class_survival = cleaned_df.groupby("Pclass")["Survived"].mean() * 100

print(class_survival)
```

```python
overall_survival = cleaned_df["Survived"].mean() * 100

print(f"Overall survival rate: {overall_survival:.2f}%")
```
